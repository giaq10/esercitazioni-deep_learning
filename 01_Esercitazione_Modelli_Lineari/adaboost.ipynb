```python
import sklearn.preprocessing
import decisionstump
import numpy as np
import matplotlib.pyplot as plt
import sklearn.tree as skt
import sklearn.ensemble as ske
```

## Load Dataset

```python
ds1 = np.genfromtxt('datasets/ds_altezzapeso.csv', delimiter=';')
(n1,d1) = ds1.shape

ds2 = np.genfromtxt('datasets/ds_altezzapeso2.csv', delimiter=';')
(n2,d2) = ds2.shape

ds = np.block([[ds1[:,:]], [ds2[:,:]]])
ds = sklearn.preprocessing.minmax_scale(ds)
y = np.block([np.ones(n1), -1*np.ones(n2)])

i = (y==1)
j = (y==-1)

clf = decisionstump.DecisionStump()
clf.fit(ds,y)


x1, y1 = clf.get_separator_2D()
print(x1,y1)

#plt.figure(1)
plt.plot(ds[i,0],ds[i,1],'.',ds[j,0],ds[j,1],'.',x1,y1,'-')
#plt.ion()
plt.show()
plt.pause(1.0001)
```

### Create Dataset

```python
np.random.seed(24)
n = 30000
x = np.random.random((n,2))
r2 = 1/(4*np.pi)
d = np.sum((x-0.5)**2, axis=1)

y = 2*(d<=r2)-1
i = np.where(y==1)[0]
j = np.where(y==-1)[0]
plt.figure()
plt.plot(x[i,0],x[i,1],'.',color='red')
plt.plot(x[j,0],x[j,1],'.',color='yellow')
plt.show()
```

```python
for i in [1,3,5,10,50,100]:
    
    ne = i
    ada_clf = ske.AdaBoostClassifier(decisionstump.DecisionStump(offset_=1/n),n_estimators=ne)
    
    # ne = 5*i
    # ada_clf = ske.AdaBoostClassifier(skt.DecisionTreeClassifier(max_depth=1),n_estimators=ne)
    #
    ada_clf.fit(x,y)
    yt = ada_clf.predict(x)
    it = (yt>0)
    jt = (yt<0)
    plt.figure()
    plt.title('Number of estimators: ' + str(ne))
    plt.plot(x[it, 0], x[it, 1], '.', color='red')
    plt.plot(x[jt, 0], x[jt, 1], '.', color='yellow')
    
    #DECISION STUMP
    for e in ada_clf.estimators_:
       xs, ys = e.get_separator_2D()
       plt.plot(xs, ys, 'k-')
    
    #DECISION TREE
    # for e in ada_clf.estimators_:
    #     feat = e.tree_.feature[0]
    #     thres = e.tree_.threshold[0]
    #     if feat < 0:            # nessuno split (foglia pura)
    #         continue
    #     if feat == 0:
    #         plt.plot([thres, thres], [0, 1], 'k-')   # verticale
    #     else:
    #         plt.plot([0, 1], [thres, thres], 'k-')   # orizzontale

    plt.show()
```
