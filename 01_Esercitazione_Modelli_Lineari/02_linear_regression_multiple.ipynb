```python
import pandas
import numpy as np
import sklearn.metrics
import sklearn.linear_model
```

## Dataset

$PLIC$: Posti letto in istituti di cura

$SC$: Spese mensili complessive

$SA/SC$: Spese alimentari / Spese complessive

$TD$: Tasso di disoccupazione

$TMI$: Tasso di mortalità infantile

```python
a = pandas.read_csv('../datasets/ds_strutture_ospedaliere.csv', sep=';', index_col = 0)
ds = np.array(a)
print("SHAPE DATASET: ", ds.shape, "\n")

target = 'TD'
x_train = a.drop(columns=target).to_numpy()
y_train = a[target].to_numpy()

print(f"X TRAIN SHAPE = {x_train.shape}")
print(f"Y TRAIN SHAPE = {y_train.shape}\n")
```

### Regressione lineare multipla

```python
lin_reg = sklearn.linear_model.LinearRegression()
lin_reg.fit(x_train, y_train)

print("Parametri:")
coef = dict(zip(a.columns.drop(target), lin_reg.coef_))
for nome, w in coef.items():
    print(f"{nome:<5}: {w:+.3f}")
print()
y_pred = lin_reg.predict(x_train)
print('MSE:',sklearn.metrics.mean_squared_error(y_train, y_pred))
```

```python
ss_res = np.sum((y_train - y_pred)**2)
ss_tot = np.sum((y_train - np.mean(y_train))**2)
r2 = 1 - ss_res / ss_tot

n, p = x_train.shape
r2_adj = 1 - (1 - r2) * (n - 1) / (n - p - 1)

print(f"{'R2:':<13}{r2:.6f}")
print(f"{'R2 (scikit):':<13}{sklearn.metrics.r2_score(y_train, y_pred):.6f}")
print(f"{'R2 corretto:':<13}{r2_adj:.6f}")
```
