```python
model = nn.Sequential(
        nn.RNN(32,16, nonlinearity='relu', batch_first=True)
        )

utils.summary(model,(1,32)) # ESEMPIO CON INPUT FORMATO DA UNA SEQUENZA LUNGA 50

#DIM = U*h + bu + V*x +  bv = (16x16)*(16x1) + (16x1) + (16x32)*(32x1) + (16x1) = (16X1)
#NUM PARAM = # DIM = U + V + b = (16x16) + (16x1) + 16x32) + (16x1) = 256 + 16 + 512 + 16 = 800
```

```python
hidden_outputs, output = model(torch.rand((100,32)))
print(hidden_outputs.shape)
print(output.shape)
print(hidden_outputs[-1,:])
print(output)
```

```python
model = nn.Sequential(
        nn.RNN(32,16, nonlinearity='relu', batch_first=True, bidirectional=True),
        )
model
utils.summary(model,(1,32))
```

```python
hidden_outputs, output = model(torch.rand((100,32)))
print(hidden_outputs.shape)
print(output.shape)
print(hidden_outputs[-1,:16])
print(hidden_outputs[0,16:])
print(output)
```
