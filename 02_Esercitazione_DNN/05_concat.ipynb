```python
import torch
import torch.nn as nn

from torch.utils.data import TensorDataset, DataLoader
```

## LOAD DATASET

```python
from datasets.datasets import load_HAR

(x_train_view_1, x_train_view_2, y_train), (x_test_view_1, x_test_view_2, y_test)  = load_HAR()

print("X TRAIN VIEW 1:",x_train_view_1.shape)
print("X TRAIN VIEW 2:",x_train_view_2.shape)
print("y TRAIN:",y_train.shape)
print()
print("X TEST VIEW 1:",x_test_view_1.shape)
print("X TEST VIEW 2:",x_test_view_2.shape)
print("y TEST:",y_test.shape)

#ETICHETTE = ["Camminare", "Salire scale", "Scendere scale", "Seduto", "In piedi", "Sdraiato"]
```

## MODEL

```python
class JointModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
                        nn.Linear(x_train_view_1.shape[1]+x_train_view_2.shape[1], 128), nn.ReLU(),
                        nn.Linear(128, 32), nn.ReLU(),
                        nn.Linear(32, 6),
                    )

    def forward(self, x_acc, x_gyro):
        return self.model(torch.cat([x_acc, x_gyro], dim=1))  # (batch, 768)


class SingleModel(nn.Module):

    def __init__(self, sensore):
        super().__init__()
        assert sensore in ("acc", "gyro")
        self.sensore = sensore
        self.model = nn.Sequential(nn.Linear(x_train_view_1.shape[1] if sensore == "acc" else x_train_view_2.shape[1], 64),
                        nn.ReLU(),
                        nn.Linear(64, 32),
                        nn.ReLU(),
                        nn.Linear(32, 6)
                    )

    def forward(self, x_acc, x_gyro):
        x = x_acc if self.sensore == "acc" else x_gyro
        return self.model(x)


class LatentConcatModel(nn.Module):

    def __init__(self):
        super().__init__()
        self.acc  = nn.Sequential(nn.Linear(x_train_view_1.shape[1], 64), nn.ReLU(), nn.Linear(64, 32), nn.ReLU())
        self.gyro = nn.Sequential(nn.Linear(x_train_view_2.shape[1], 64), nn.ReLU(), nn.Linear(64, 32), nn.ReLU())
        self.testa = nn.Sequential(nn.Linear(2 * 32, 6))

    def forward(self, x_acc, x_gyro):
        z = torch.cat([self.acc(x_acc), self.gyro(x_gyro)], dim=1)  # (batch, 64)
        return self.testa(z)

COSTRUTTORI = {
    "Solo accelerometro":  lambda: SingleModel("acc"),
    "Solo giroscopio":     lambda: SingleModel("gyro"),
    "Early fusion":        lambda: JointModel(),
    "Intermediate fusion": lambda: LatentConcatModel(),
}

def conta_parametri(m):
    return sum(p.numel() for p in m.parameters())

for nome, crea in COSTRUTTORI.items():
    print(f"{nome:<22} {conta_parametri(crea()):>8,} parametri")
```

## Training

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

CLASSI = sorted(y_train.unique())
classe2idx = {c: i for i, c in enumerate(CLASSI)}
print(classe2idx)

def to_tensor(df):
    return torch.tensor(df.to_numpy(), dtype=torch.float32)

Xtr_acc, Xtr_gyro = to_tensor(x_train_view_1), to_tensor(x_train_view_2)
Xte_acc, Xte_gyro = to_tensor(x_test_view_1), to_tensor(x_test_view_2)
ytr = torch.tensor(y_train.map(classe2idx).to_numpy(), dtype=torch.long)
yte = torch.tensor(y_test.map(classe2idx).to_numpy(), dtype=torch.long)

train_ds = TensorDataset(Xtr_acc, Xtr_gyro, ytr)
test_ds  = TensorDataset(Xte_acc, Xte_gyro, yte)
train_eval_loader = DataLoader(train_ds, batch_size=1024)
test_loader       = DataLoader(test_ds,  batch_size=1024)
```

```python
EPOCHE, BATCH_SIZE, LR = 30, 64, 1e-3

def valuta(model, loader):
    model.eval()
    pred, vere = [], []
    with torch.no_grad():
        for xa, xg, y in loader:
            pred.append(model(xa.to(device), xg.to(device)).argmax(dim=1).cpu())
            vere.append(y)
    pred, vere = torch.cat(pred), torch.cat(vere)
    return (pred == vere).float().mean().item(), pred.numpy(), vere.numpy()

def addestra(crea_modello):
    model = crea_modello().to(device)
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR)
    loss_fn = nn.CrossEntropyLoss()
    storia = []
    for epoca in range(EPOCHE):
        model.train()
        somma = 0.0
        for xa, xg, y in train_loader:
            xa, xg, y = xa.to(device), xg.to(device), y.to(device)
            optimizer.zero_grad()
            loss = loss_fn(model(xa, xg), y)
            loss.backward()
            optimizer.step()
            somma += loss.item() * len(y)
        storia.append(somma / len(train_ds))
    return model, storia
```

```python
for nome, crea in COSTRUTTORI.items():
    model, storia = addestra(crea)

    acc_train, _, _ = valuta(model, train_eval_loader)
    acc_test,  _, _ = valuta(model, test_loader)

    print(f"{nome:<22} | train {acc_train:.4f} | test {acc_test:.4f}")
```
